б) Сколько истинных (не объясняемых случайностью) «горбов» 
(локальных максимумов) имеет распределение признака?

Видим 2 "горба": в начале и где-то между 2 и 3.

в)

```python
from scipy.stats import trim_mean
import numpy as np

d_new = d[(d['LogFrequency']!=0) & (d['SemanticClass']=='opaque')]
f = d_new['LogFrequency']
mean = f.mean()
trim = trim_mean(f, 0.05)
H = np.quantile(f, 0.75)-np.quantile(f, 0.25)
print('Среднее=',round(mean,2))
print('Усеченное среднее=',round(trim,2))
print('Размах=',round(H,2))
```

Вывод:

Распределение трудных слов без нулевых частот практически не имеет выбросов (усеченное среднее ≈ среднее). При это размах 1.31, т.е. значения группируются вокруг среднего. Это соответсвует правому "горбу" нашей гистограммы.
