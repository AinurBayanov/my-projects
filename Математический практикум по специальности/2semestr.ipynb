```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind, mannwhitneyu

def power_at_mu_normal(mu, n1, n2, alpha=0.05, n_sim=500):
    reject_t_equal = 0
    reject_t_perm = 0
    reject_mw_exact = 0
    reject_mw_asymp = 0
    
    for _ in range(n_sim):
        x = np.random.normal(0, 1, n1)
        y = np.random.normal(mu, 1, n2)
        
        _, p_eq = ttest_ind(x, y, equal_var=True)
        if p_eq < alpha:
            reject_t_equal += 1
            
        res_perm = ttest_ind(x, y, permutations=100, random_state=None)
        if res_perm.pvalue < alpha:
            reject_t_perm += 1
        
        
        _, p_mw_exact = mannwhitneyu(x, y, method='exact')
        if p_mw_exact < alpha:
            reject_mw_exact += 1
            
        _, p_mw_asymp = mannwhitneyu(x, y, method='asymptotic')
        if p_mw_asymp < alpha:
            reject_mw_asymp += 1
    
    return {
        't_equal_var': reject_t_equal / n_sim,
        't_perm': reject_t_perm / n_sim,
        'mw_exact': reject_mw_exact / n_sim,
        'mw_asymp': reject_mw_asymp / n_sim
    }

def plot_power_curves_normal(mu_range, n1, n2, ax):
    powers = []
    for mu in mu_range:
        pwr = power_at_mu_normal(mu, n1, n2, n_sim=300)
        powers.append(pwr)
    
    t_eq = [p['t_equal_var'] for p in powers]
    t_perm = [p['t_perm'] for p in powers]
    mw_ex = [p['mw_exact'] for p in powers]
    mw_as = [p['mw_asymp'] for p in powers]
    
    ax.plot(mu_range, t_eq, '-', color = 'black', label='t-test (equal var)', linewidth=2)
    ax.plot(mu_range, t_perm, '-', color = 'red', label='t-test (permutation)', linewidth=2)
    ax.plot(mu_range, mw_ex, '-',color ='blue', label='Mann-Whitney (exact)', linewidth=3)
    ax.plot(mu_range, mw_as, '-', color ='green', label='Mann-Whitney (asymptotic)', linewidth=2)
    ax.legend(fontsize=10, loc='lower right')
    ax.grid(True, alpha=0.3)

n_sizes = [10, 50, 100]
mu_range_norm = np.linspace(0, 2, 20)

fig, axes = plt.subplots(3, 1, figsize=(8, 12)) 

for i, n in enumerate(n_sizes):
    plot_power_curves_normal(mu_range_norm, n, n, axes[i])

plt.suptitle('Мощность критериев для нормального распределения', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import mannwhitneyu, ks_2samp, anderson_ksamp, t

ad_values = []
with open("Anderson_Darling.txt", "r") as f:
    for line in f:
        ad_values.append(float(line.strip()))
ad_values = np.array(ad_values)
N = len(ad_values)

def bws_statistic(x, y):
    n1 = len(x)
    n2 = len(y)
    n = n1 + n2

    combined = np.concatenate([x, y])
    ranks = np.argsort(np.argsort(combined)) + 1
  
    ranks_x = np.sort(ranks[:n1])
    
    ranks_y = np.sort(ranks[n1:])
    
    sum1 = 0.0
    for i in range(1, n1 + 1):
        r = ranks_x[i - 1]
        p = i / (n1 + 1)
        denominator = p * (1 - p) * (n2 * n / n1)
        if denominator > 0:
            term = ((r - (n * i / n1)) ** 2) / denominator
            sum1 += term
    
    sum2 = 0.0
    for i in range(1, n2 + 1):
        r = ranks_y[i - 1]
        p = i / (n2 + 1)
        denominator = p * (1 - p) * (n1 * n / n2)
        if denominator > 0:
            term = ((r - (n * i / n2)) ** 2) / denominator
            sum2 += term
    
    bws = (sum1 / (2 * n1)) + (sum2 / (2 * n2))
    return bws

def get_pvalue(x, y):
    T = bws_statistic(x, y)
    count_less = np.sum(ad_values < T)
    pvalue = 1 - count_less / N
    return pvalue

def generate_data(case, n1=50, n2=50, seed=None):
    if seed is not None:
        np.random.seed(seed)
    if case == 'a':
        # N(0,1) и N(0,1)
        x = np.random.normal(0, 1, n1)
        y = np.random.normal(0, 1, n2)
    elif case == 'b':
        # N(0,1) и N(0.3,1)
        x = np.random.normal(0, 1, n1)
        y = np.random.normal(0.3, 1, n2)
    elif case == 'c':
        # N(0,1) и N(0, 3/2)
        x = np.random.normal(0, 1, n1)
        y = np.random.normal(0, np.sqrt(1.5), n2)
    elif case == 'd':
        # N(0,1) и t_k  (k=3)
        x = np.random.normal(0, 1, n1)
        y = t.rvs(df=3, size=n2)
    elif case == 'e':
        # N(0,1) и центрированная нормированная сумма k=3 из R[-1,1]
        k = 3
        y_raw = np.sum(np.random.uniform(-1, 1, (n2, k)), axis=1)
        y = y_raw * (3 ** (0.5))
        x = np.random.normal(0, 1, n1)
    else:
        raise ValueError("Unknown case")
    return x, y

def ecdf_pvalues(pvals):
    pvals_sorted = np.sort(pvals)
    ecdf = np.arange(1, len(pvals_sorted) + 1) / len(pvals_sorted)
    return pvals_sorted, ecdf



def compare_criteria(case, n1=50, n2=50, n_sim=300):
    pvals_mw = []
    pvals_ks = []
    pvals_ad = []
    pvals_bws = []

    for i in range(n_sim):
        x, y = generate_data(case, n1, n2, seed=i)

        # Mann-Whitney
        _, p_mw = mannwhitneyu(x, y, method='asymptotic')
        pvals_mw.append(p_mw)

        # Kolmogorov-Smirnov
        _, p_ks = ks_2samp(x, y)
        pvals_ks.append(p_ks)

        # Anderson-Darling
        res_ad = anderson_ksamp([x, y])
        pvals_ad.append(res_ad.pvalue)

        # BWS
        p_bws = get_pvalue(x, y)
        pvals_bws.append(p_bws)

    plt.figure(figsize=(8, 6))

    for pvals, label, color in zip(
        [pvals_mw, pvals_ks, pvals_ad, pvals_bws],
        ['Mann-Whitney', 'Kolmogorov-Smirnov', 'Anderson-Darling', 'BWS'],
        ['blue', 'green', 'red', 'purple'],
    ):
        sorted_p, ecdf = ecdf_pvalues(pvals)
        plt.step(sorted_p, ecdf, color=color, linewidth=2, label=label)

    plt.plot([0, 1], [0, 1], 'k--', alpha=0.5)
    plt.legend(loc='lower right')
    plt.grid(True, alpha=0.3)
    plt.xlim(0, 1)
    plt.ylim(0, 1)
    plt.tight_layout()
    plt.show()



for case in ['a', 'b', 'c', 'd', 'e']:
    compare_criteria(case, n1=50, n2=50, n_sim=300)
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = 10000
num_simulations = 5000

x_theory = np.linspace(0.001, 0.999, 1000)
f_arcsine = 1 / (np.pi * np.sqrt(x_theory * (1 - x_theory)))


def generate_pssb(n):
    steps = np.random.choice([-1, 1], size=n)
    return np.cumsum(steps)

def generate_sb0(n):
    steps = np.random.normal(0, 1, size=n)
    return np.cumsum(steps)

def generate_cauchy_sb(n):
    steps = np.random.standard_cauchy(size=n)
    return np.cumsum(steps)


def last_zero_crossing_index(traj):
    signs = np.sign(traj)
    changes = np.where(np.diff(signs) != 0)[0]
    
    if len(changes) == 0:
        return 0
    last_cross_idx = changes[-1] + 1
    
    return last_cross_idx

def positive_time_fraction(traj):
    return np.sum(traj[1:] > 0) / len(traj[1:])


results_pssb_last_cross = []
results_pssb_pos_time = []

results_sb0_last_cross = []
results_sb0_pos_time = []

results_cauchy_last_cross = []
results_cauchy_pos_time = []

for _ in range(num_simulations):
    traj_pssb = generate_pssb(n)
    traj_pssb = np.insert(traj_pssb, 0, 0) 
    idx_cross_pssb = last_zero_crossing_index(traj_pssb)
    results_pssb_last_cross.append(idx_cross_pssb / n)
    results_pssb_pos_time.append(positive_time_fraction(traj_pssb))

    traj_sb0 = generate_sb0(n)
    traj_sb0 = np.insert(traj_sb0, 0, 0)
    idx_cross_sb0 = last_zero_crossing_index(traj_sb0)
    results_sb0_last_cross.append(idx_cross_sb0 / n)
    results_sb0_pos_time.append(positive_time_fraction(traj_sb0))

    traj_cauchy = generate_cauchy_sb(n)
    traj_cauchy = np.insert(traj_cauchy, 0, 0)
    idx_cross_cauchy = last_zero_crossing_index(traj_cauchy)
    results_cauchy_last_cross.append(idx_cross_cauchy / n)
    results_cauchy_pos_time.append(positive_time_fraction(traj_cauchy))


results_pssb_last_cross = np.array(results_pssb_last_cross)
results_pssb_pos_time = np.array(results_pssb_pos_time)

results_sb0_last_cross = np.array(results_sb0_last_cross)
results_sb0_pos_time = np.array(results_sb0_pos_time)

results_cauchy_last_cross = np.array(results_cauchy_last_cross)
results_cauchy_pos_time = np.array(results_cauchy_pos_time)


fig, axes = plt.subplots(2, 3, figsize=(18, 10))


titles_col = ['ПССБ', 'СБ0 $N(0,1)$', 'Коши']
titles_row_a = '1(a): Момент последнего пересечения нуля'
titles_row_b = '1(b): Доля времени на положительной полуплоскости'

colors = ['blue', 'green', 'orange']

for i, (data, color, title) in enumerate(zip(
    [results_pssb_last_cross, results_sb0_last_cross, results_cauchy_last_cross],
    colors, titles_col)):
    
    ax = axes[0, i]
    ax.hist(data, bins=50, density=True, alpha=0.7, color=color, label='Гистограмма')
    ax.plot(x_theory, f_arcsine, 'r-', linewidth=2, label='Плотность')
    ax.set_title(f'{title}\n{titles_row_a}')
    ax.legend()

for i, (data, color, title) in enumerate(zip(
    [results_pssb_pos_time, results_sb0_pos_time, results_cauchy_pos_time],
    colors, titles_col)):
    
    ax = axes[1, i]
    ax.hist(data, bins=50, density=True, alpha=0.7, color=color, label='Гистограмма')
    ax.plot(x_theory, f_arcsine, 'r-', linewidth=2, label='Плотность')
    ax.set_title(f'{title}\n{titles_row_b}')
    ax.legend()

plt.tight_layout()
plt.savefig('arcsine_law_crossing_zero.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import eig
from scipy.stats import norm

def simulate_markov_chain(P, n_steps):
    n_states = 3
    trajectory = [0]
    current_state = 0
    for _ in range(n_steps - 1):
        next_state = np.random.choice(n_states, p=P[current_state])
        trajectory.append(next_state)
        current_state = next_state
    return trajectory

def compute_visit_fractions(trajectory, n_states):
    counts = np.zeros(n_states)
    for state in trajectory:
        counts[state] += 1
    fractions = counts / len(trajectory)
    return fractions

P = np.array([
    [1/3, 1/3, 1/3],
    [3/4, 0,   1/4],
    [1/2, 1/6, 1/3]
])

#pi = [9/19, 4/19, 6/19]

n_trajectories = 100
n_steps_centrallimit = 1000

normalized_deviations = []
for i in range(n_trajectories):
    trajectory = simulate_markov_chain(P, n_steps_centrallimit)
    fractions = compute_visit_fractions(trajectory, 3)
    fraction_state_0 = fractions[0]
    deviation = fraction_state_0 - 9/19
    normalized_deviation = deviation * np.sqrt(n_steps_centrallimit)
    normalized_deviations.append(normalized_deviation)

normalized_deviations = np.array(normalized_deviations)

n_steps_plot = 50 
trajectory_plot = simulate_markov_chain(P, n_steps_plot)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].hist(normalized_deviations, bins=20, density=True, alpha=0.7, 
             color='skyblue', edgecolor='black')
axes[0].grid(True, alpha=0.3)

x_norm = np.linspace(min(normalized_deviations) - 1, max(normalized_deviations) + 1, 100)
sigma_empirical = np.std(normalized_deviations)
axes[0].plot(x_norm, norm.pdf(x_norm, 0, sigma_empirical), 'r-', linewidth=2, 
             label=f'N(0, σ²={sigma_empirical**2:.4f})')
axes[0].legend(fontsize=9)

time = np.arange(len(trajectory_plot))
axes[1].plot(time, trajectory_plot, marker='o', markersize=5, linestyle='-', color='darkblue', linewidth=1.5)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_excess_life(step_dist, lambda_param=None, T=1000, n_simulations=10000):
    excesses = []
    for _ in range(n_simulations):
        S = 0
        while S <= T:
            if step_dist == 'exp':
                step = np.random.exponential(scale=1/lambda_param)
            elif step_dist == 'uniform':
                step = np.random.uniform(0, 1)
            else:
                raise ValueError("Unknown distribution")
            S += step
        # S теперь > T, предыдущий шаг был до T
        excess = S - T
        excesses.append(excess)
    return np.array(excesses)

# Часть а): Exp(1)
np.random.seed(42)
excess_exp = simulate_excess_life('exp', lambda_param=1, T=1000, n_simulations=10000)

# Часть б): Uniform[0,1]
excess_unif = simulate_excess_life('uniform', T=1000, n_simulations=10000)

# Построение графиков
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# a) Exp(1)
axes[0].hist(excess_exp, bins=50, density=True, alpha=0.7, color='blue', label='Симуляция')
x = np.linspace(0, 5, 1000)
axes[0].plot(x, np.exp(-x), 'r-', linewidth=2, label='Теория: Exp(1)')
axes[0].set_title('Экспцесс для Exp(1) шагов')
axes[0].legend()
axes[0].set_xlabel('Время')
axes[0].set_ylabel('Плотность')

# b) Uniform[0,1]
axes[1].hist(excess_unif, bins=50, density=True, alpha=0.7, color='green', label='Симуляция')
x_unif = np.linspace(0, 1, 1000)
axes[1].plot(x_unif, 2*(1 - x_unif), 'r-', linewidth=2, label='Теория: 2(1-x)')
axes[1].set_title('Экспцесс для Uniform[0,1] шагов')
axes[1].legend()
axes[1].set_xlabel('Время')
axes[1].set_ylabel('Плотность')

plt.tight_layout()
plt.show()
```

```python
import numpy as np

def read_transition_matrix(filename):
    with open(filename, 'r') as f:
        lines = f.readlines()
    
    n = len(lines)
    P = np.zeros((n, n))
    
    for i, line in enumerate(lines):
        values = [float(x) for x in line.strip().split(';')]
        P[i, :len(values)] = values
    
    return P

def find_reachable_states(P, start_state):
    n = P.shape[0]
    visited = set()
    queue = [start_state]
    visited.add(start_state)
    
    while queue:
        current = queue.pop(0)
        for next_state in range(n):
            if P[current, next_state] > 1e-10 and next_state not in visited:
                visited.add(next_state)
                queue.append(next_state)
    
    return visited

def find_communicating_classes(P):
    n = P.shape[0]
    visited = set()
    classes = []
    
    for state in range(n):
        if state not in visited:
            reachable_from_state = find_reachable_states(P, state)
            
            communicating_class = set()
            for other_state in reachable_from_state:
                reachable_from_other = find_reachable_states(P, other_state)
                if state in reachable_from_other:
                    communicating_class.add(other_state)
            
            if communicating_class:
                classes.append(communicating_class)
                visited.update(communicating_class)
    
    return classes

P = read_transition_matrix('MarkovChain.txt')
classes = find_communicating_classes(P)

print(f"Найдено {len(classes)} классов:")
for i, cls in enumerate(classes, 1):
    print(f"Класс {i}: {sorted(list(cls))}")
```
